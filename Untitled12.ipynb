```python
print("--- Banco ----")
dinheiro = 1067
nome = input("Digite seu nome: ")
print(f"Bem vindo(a), {nome}!")

print("\nQual operação você deseja fazer?")
print("1 - Saque")
print("2 - Deposito")
print("3 - Extrato")

opcao = int(input("Escolha uma opção (1-3): "))

if opcao == 1:
    valor = float(input("Digite o valor a ser sacado: "))
    if valor <= dinheiro:
        dinheiro -= valor
        print(f"Saque de R$ {valor:.2f} realizado com sucesso!")
    else:
        print("Saldo insuficiente para realizar o saque.")

elif opcao == 2:
    valor = float(input("Digite o valor a ser depositado: "))
    dinheiro += valor
    print(f"Depósito de R$ {valor:.2f} realizado com sucesso!")

elif opcao == 3:
    print(f"Seu extrato atual é: R$ {dinheiro:.2f}")

else:
    print("Opção inválida!")
```

```python
import random

tarot_deck = {
    "O Louco": {
        "reto": "Novos começos, espontaneidade, liberdade.",
        "invertido": "Imprudência, risco desnecessário, hesitação.",
    },
    "O Mago": {
        "reto": "Habilidade, poder pessoal, concentração, ação.",
        "invertido": "Manipulação, falta de planejamento, talentos desperdiçados.",
    },
    "A Sacerdotisa": {
        "reto": "Intuição, mistério, sabedoria interior, inconsciente.",
        "invertido": "Segredos revelados, desconexão com a intuição.",
    },
    "A Imperatriz": {
        "reto": "Abundância, natureza, criatividade, cuidado.",
        "invertido": "Dependência, bloqueio criativo, excessos.",
    },
    "O Imperador": {
        "reto": "Autoridade, estrutura, controle, estabilidade.",
        "invertido": "Tirania, rigidez, falta de disciplina.",
    },
    "A Morte": {
        "reto": "Transformação profunda, fim de um ciclo, renascimento.",
        "invertido": "Resistência à mudança, estagnação, medo do novo.",
    },
    "O Sol": {
        "reto": "Alegria, sucesso, vitalidade, clareza.",
        "invertido": "Tristeza temporária, otimismo excessivo.",
    },
}

def tirar_cartas():

    cartas_disponiveis = list(tarot_deck.keys())


    sorteadas = random.sample(cartas_disponiveis, 3)

    posicoes = ["Passado", "Presente", "Futuro"]

    print("\n🔮 --- SUA LEITURA DE TAROT --- 🔮\n")
    print("Faça sua pergunta mentalmente")
    print("Cartas sorteadas:")

    for posicao, carta in zip(posicoes, sorteadas):
        orientacao = random.choice(["reto", "invertido"])
        significado = tarot_deck[carta][orientacao]

        print(f"[{posicao}]: {carta} ({orientacao.upper()})")
        print(f"   -> Significado: {significado}\n")

if __name__ == "__main__":
    tirar_cartas()
```

```python
from IPython.display import HTML
jogo_cobrinha = """
<div style="text-align: center; font-family: Arial, sans-serif; background-color: #ADD8E6; padding: 20px; border-radius: 10px; width: 430px; margin: 0 auto;">
    <h2 style="color: #fff; margin-bottom: 10px;">Jogo da Cobrinha 67</h2>
    <div style="color: #fff; margin-bottom: 10px; font-size: 18px;">Pontuação: <span id="score">0</span></div>
    <canvas id="gameCanvas" width="400" height="400" style="background-color: #0000FF; border: 2px solid #555; display: block; margin: 0 auto;"></canvas>
    <p style="color: #aaa; font-size: 12px; margin-top: 10px;">Use as **setinhas** para jogar.</p>
    <button id="restartBtn" style="padding: 8px 15px; font-size: 14px; background-color: #4CAF50; color: white; border: none; border-radius: 5px; cursor: pointer; display: none; margin: 10px auto 0;">Jogar Novamente</button>
</div>

<script>
    const canvas = document.getElementById("gameCanvas");
    const ctx = canvas.getContext("2d");
    const scoreElement = document.getElementById("score");
    const restartBtn = document.getElementById("restartBtn");

    const gridSize = 20;
    const tileCount = canvas.width / gridSize;

    let snake, dx, dy, food, score, gameInterval;

    function initGame() {
        snake = [
            {x: 10, y: 10},
            {x: 9, y: 10},
            {x: 8, y: 10}
        ];
        dx = 1;
        dy = 0;
        score = 0;
        scoreElement.innerText = score;
        restartBtn.style.display = "none";
        resetFood();
        if (gameInterval) clearInterval(gameInterval);
        gameInterval = setInterval(update, 100);
    }

    function resetFood() {
        food = {
            x: Math.floor(Math.random() * tileCount),
            y: Math.floor(Math.random() * tileCount)
        };
        snake.forEach(part => {
            if (part.x === food.x && part.y === food.y) resetFood();
        });
    }

    function update() {
        const head = {x: snake[0].x + dx, y: snake[0].y + dy};
        if (head.x < 0 || head.x >= tileCount || head.y < 0 || head.y >= tileCount || collisionCheck(head)) {
            clearInterval(gameInterval);
            ctx.fillStyle = "rgba(0, 0, 0, 0.75)";
            ctx.fillRect(0, 0, canvas.width, canvas.height);
            ctx.fillStyle = "red";
            ctx.font = "30px Arial";
            ctx.textAlign = "center";
            ctx.fillText("Fim de jogo :( ", canvas.width / 2, canvas.height / 2);
            restartBtn.style.display = "block";
            return;
        }

        snake.unshift(head);
        if (head.x === food.x && head.y === food.y) {
            score += 100;
            scoreElement.innerText = score;
            resetFood();
        } else {
            snake.pop();
        }

        draw();
    }

    function collisionCheck(head) {
        return snake.some((part, index) => index !== 0 && part.x === head.x && part.y === head.y);
    }

    function draw() {
        ctx.fillStyle = "#111";
        ctx.fillRect(0, 0, canvas.width, canvas.height);


        snake.forEach((part, index) => {
            ctx.fillStyle = index === 0 ? "#4CAF50" : "#81C784"; // Cabeça mais escura
            ctx.fillRect(part.x * gridSize, part.y * gridSize, gridSize - 2, gridSize - 2);
        });


        ctx.fillStyle = "#FF5722";
        ctx.fillRect(food.x * gridSize, food.y * gridSize, gridSize - 2, gridSize - 2);
    }

    window.addEventListener("keydown", e => {
        switch(e.key) {
            case "ArrowUp":
                if (dy === 0) { dx = 0; dy = -1; }
                e.preventDefault();
                break;
            case "ArrowDown":
                if (dy === 0) { dx = 0; dy = 1; }
                e.preventDefault();
                break;
            case "ArrowLeft":
                if (dx === 0) { dx = -1; dy = 0; }
                e.preventDefault();
                break;
            case "ArrowRight":
                if (dx === 0) { dx = 1; dy = 0; }
                e.preventDefault();
                break;
        }
    });

    restartBtn.addEventListener("click", initGame);


    initGame();
</script>
"""
HTML(jogo_cobrinha)
```
