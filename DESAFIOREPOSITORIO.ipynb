```python
import requests
from bs4 import BeautifulSoup

url_books = "https://books.toscrape.com/"
palavras_livros = ["light", "star", "life", "secret"]

try:
    pagina = requests.get(url_books)
    pagina.raise_for_status()
except Exception as erro:
    print("Erro ao acessar o site de livros:", erro)
else:
    soup = BeautifulSoup(pagina.content, "html.parser")

    titulos_h3 = soup.find_all("h3")

    print("=== LIVROS ENCONTRADOS ===")
    for h3 in titulos_h3:
        tag_a = h3.find("a")

        titulo = tag_a["title"]

        for palavra in palavras_livros:
            if palavra.lower() in titulo.lower():
                print(f"Palavra encontrada: '{palavra}'")
                print(f"Título do livro: {titulo}\n")
                break
```
