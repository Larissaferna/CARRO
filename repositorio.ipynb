```python
import requests
from bs4 import BeautifulSoup

palavras = ["love", "abilities", "success", "inspirational"]
citacoes_encontradas = 0

for pagina_num in range(1, 3):
    url = f"https://quotes.toscrape.com/page/{pagina_num}/"

    try:
        pagina = requests.get(url)
        pagina.raise_for_status()
    except Exception as erro:
        print(f"Erro ao acessar a página {pagina_num}:", erro)
        continue

    soup = BeautifulSoup(pagina.content, "html.parser")
    citacoes = soup.find_all("div", class_="quote")

    for citacao in citacoes:
        texto = citacao.find("span", class_="text").get_text()
        autor = citacao.find("small", class_="author").get_text()

        if any(palavra.lower() in texto.lower() for palavra in palavras):
            citacoes_encontradas += 1

            div_tags = citacao.find("div", class_="tags")
            tags_elementos = div_tags.find_all("a", class_="tag")
            tags = [tag.get_text() for tag in tags_elementos]

            print(f"Texto: {texto}")
            print(f"Autor: {autor}")
            print(f"Tags: {', '.join(tags)}")
            print("-" * 50)

print(f"\nTotal de citações encontradas: {citacoes_encontradas}")
```
